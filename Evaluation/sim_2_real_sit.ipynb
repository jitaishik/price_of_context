```python
import json
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics import (
    roc_curve, auc,
    accuracy_score, f1_score,
    precision_score, recall_score,
    confusion_matrix
)
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
with open("../valid_indices.json", "r") as f:
    valid_indices = json.load(f)
```

```python
with open('path_to_situation_extraction', 'r') as file:
    situation_1 = json.load(file)
```

```python
df = pd.read_parquet("../eeyore-data.parquet")
```

```python
profiles = df['profile'].tolist()
```

```python
situation_1_list = []
situation_2_list = []
session_list = []
```

```python
for i in range(550):
    if "session_"+str(i) in situation_1.keys() and i-1 in valid_indices:
        profile_i = json.loads(profiles[i-1])
        situation_1_list.append(situation_1["session_"+str(i)])
        situation_2_list.append(profile_i["situation of the client"])
        session_list.append("session_"+str(i))
```

```python
embeddings_situation_1 = model.encode(situation_1_list)
```

```python
embeddings_situation_2 = model.encode(situation_2_list)
```

```python
embeddings1 = normalize(embeddings_situation_1)
embeddings2 = normalize(embeddings_situation_2)
```

```python
similarity_matrix = np.matmul(embeddings1, embeddings2.T)
```
