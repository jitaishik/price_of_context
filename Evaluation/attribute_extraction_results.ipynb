```python
import pandas as pd
import json
import os
import re
import ast
```

```python
marital_status_mapping = {'Currently experiencing a breakup with her husband.':'Separated','Divorced':'Divorced','Engaged':'In a relationship','In a relationship':'In a relationship',
                         'In a relationship, unmarried':'In a relationship','Married':'Married','Not Married':'Other','Separated':'Separated','Single':'Single','Unmarried':'Other','Widowed':'Widowed','widowed':'Widowed','single':'Single','Not married':'Other'}
```

```python
with open("../valid_indices.json", "r") as f:
    valid_indices = json.load(f)
```

```python
def clean_value(x):
    # Convert non-strings to string
    if not isinstance(x, str):
        return str(x)

    x = x.strip()
    if x.startswith('[') and x.endswith(']'):
        try:
            parsed = ast.literal_eval(x)
            if isinstance(parsed, list) and parsed:
                return str(parsed[0]).strip("'\"")
        except (ValueError, SyntaxError):
            pass
    return x.strip("'\"")
```

```python
tp=0
fp=0
tn=0
fn=0
```

```python
for filename in os.listdir("path_to_extracted_demographics"):
    if filename.endswith('.json'):
        x = int(filename.replace('session_', '').replace('.json', ''))
        if (x-1) in valid_indices:
            with open("path_to_extracted_demographics" + filename) as f:
                data = json.load(f)
            for key in data['synthetic profile'].keys():
                if clean_value(data['synthetic profile'][key]) == 'Cannot be identified':
                    if key not in data['actual profile'].keys():
                        tn = tn+1
                    else:
                        fn = fn+1
                else:
                    if key not in data['actual profile'].keys():
                        fp = fp+1
                    else:
                        if key =='marital status':
                            data['actual profile'][key] = marital_status_mapping[data['actual profile'][key]]
                        if data['actual profile'][key].lower() == clean_value(data['synthetic profile'][key]).lower():
                            tp = tp+1
                            print(filename)
                            print(data['actual profile'][key])
                        else:
                            fp = fp+1
```
